```python
!nslookup google.com
```

- ICMP(Ping)은 차단되어도, 웹 검색을 위한 DNS는 통신이 열려있는 경우가 많다.
- 이를 통해 도메인 이름이 IP로 변환되는 과정을 확인할 수 있다.

```python
!nslookup -type=MX google.com
```

- cns3.bora.net은 LG유플러스가 운영하는 전용선을 사용하고 있음을 알 수 있다.
- 내 PC가 도메인 이름인 goole.com을 IP로 바꾸기 위해서 요청을 보낸 DNS 서버가 LG유플러스의 공용 DNS 서버임을 알 수 있다.
- 이때, DNS 서버의 IP는 203.248.252.2이다.

- MX는 Mail Exchanger로, 메일 수신 서버 정보를 나타내는 유형이다.
- smtp.google.com가 실제로 메일을 받아 처리하는 서버 정보이다.
- MX preference가 10이라는 것은 우선순위 10번으로 숫자가 낮을수록 우선권이 높으며, 구글은 기본 10번 서버를 주 메일 서버로 지정해두었다.

- 그 아래 Non-authoritative answer:의 의미는 LG유플러스 DNS 서버인 cns3.bora.net이 내가 핑을 보낸 google.com의 주인이 아니라, 중간에서 대신 물어보고 전해준 응답이라는 의미이다. (따라서 권한이 없는 응답이라는 뜻)
